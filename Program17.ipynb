```python

import pandas as pd
from sklearn.linear_model import LinearRegression

# Create dataset
data = {
    'RAM': [2, 3, 4, 6, 8, 12],
    'Storage': [32, 32, 64, 128, 256, 512],
    'Price': [5000, 7000, 10000, 15000, 20000, 30000]
}

# Convert into DataFrame
df = pd.DataFrame(data)

# Input and Output
X = df[['RAM', 'Storage']]
y = df['Price']

# Train the model
model = LinearRegression()
model.fit(X, y)

# Predict mobile price
ram = 8
storage = 256

prediction = model.predict([[ram, storage]])

print("Mobile RAM:", ram, "GB")
print("Mobile Storage:", storage, "GB")
print("Predicted Mobile Price: Rs.", round(prediction[0], 2))
```
