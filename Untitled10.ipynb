```python

import numpy as np
import matplotlib.pyplot as plt
from sklearn.mixture import GaussianMixture

# Generate sample data
np.random.seed(42)
X1 = np.random.normal(5, 1, 100)
X2 = np.random.normal(10, 1, 100)
X = np.concatenate([X1, X2]).reshape(-1, 1)

# Apply EM Algorithm
gmm = GaussianMixture(n_components=2, random_state=42)
gmm.fit(X)

# Predict clusters
labels = gmm.predict(X)

# Print output
print("Cluster Means:")
print(gmm.means_)

print("\nCluster Weights:")
print(gmm.weights_)

print("\nEM Algorithm Completed Successfully")

# Plot clusters
plt.scatter(X, np.zeros_like(X), c=labels, cmap='viridis')
plt.title("Expectation Maximization Algorithm")
plt.xlabel("Data Values")
plt.ylabel("Clusters")
plt.grid()
plt.show()
```
