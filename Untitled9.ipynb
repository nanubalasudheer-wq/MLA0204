```python

import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures

# Dataset
X = np.array([1, 2, 3, 4, 5, 6]).reshape(-1, 1)
y = np.array([2, 5, 10, 17, 26, 37])

# Linear Regression
linear_model = LinearRegression()
linear_model.fit(X, y)
y_linear = linear_model.predict(X)

# Polynomial Regression
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

poly_model = LinearRegression()
poly_model.fit(X_poly, y)

# Smooth curve for plotting
X_curve = np.linspace(1, 6, 100).reshape(-1, 1)
X_curve_poly = poly.transform(X_curve)
y_poly = poly_model.predict(X_curve_poly)
y_curve_linear = linear_model.predict(X_curve)

# Display two graphs
plt.figure(figsize=(12, 5))

# Linear Regression Graph
plt.subplot(1, 2, 1)
plt.scatter(X, y, color='blue', label='Actual Data')
plt.plot(X_curve, y_curve_linear, color='red', label='Linear Line')
plt.title('Linear Regression')
plt.xlabel('X Values')
plt.ylabel('Y Values')
plt.legend()
plt.grid()

# Polynomial Regression Graph
plt.subplot(1, 2, 2)
plt.scatter(X, y, color='blue', label='Actual Data')
plt.plot(X_curve, y_poly, color='green', label='Polynomial Curve')
plt.title('Polynomial Regression')
plt.xlabel('X Values')
plt.ylabel('Y Values')
plt.legend()
plt.grid()

plt.tight_layout()
plt.show()
```
