```python

import pandas as pd
from sklearn.naive_bayes import GaussianNB
from sklearn.preprocessing import LabelEncoder

# Create dataset
data = {
    'Income': [20000, 25000, 30000, 40000, 50000,
               60000, 70000, 80000, 35000, 45000],
    'Credit_Score': [500, 550, 600, 650, 700,
                     750, 800, 850, 580, 680],
    'Loan_Status': ['No', 'No', 'No', 'Yes', 'Yes',
                    'Yes', 'Yes', 'Yes', 'No', 'Yes']
}

df = pd.DataFrame(data)

# Input and output
X = df[['Income', 'Credit_Score']]
y = df['Loan_Status']

# Train model
model = GaussianNB()
model.fit(X, y)

# Test new customer
new_customer = [[45000, 700]]

prediction = model.predict(new_customer)

print("Bank Loan Prediction")
print("Income:", new_customer[0][0])
print("Credit Score:", new_customer[0][1])
print("Loan Status:", prediction[0])
```
